```python
import pandas as pd

data = {
    'Order_ID': [101, 102, 103, 104, 105, 106,
                 107, 108, 109, 110, 111, 112],

    'Customer': ['Amit', 'Neha', 'Rahul', 'Priya',
                 'Amit', 'Neha', 'Karan', 'Priya',
                 'Rahul', 'Karan', 'Amit', 'Neha'],

    'City': ['Delhi', 'Mathura', 'Delhi', 'Agra',
             'Mathura', 'Delhi', 'Agra', 'Mathura',
             'Agra', 'Delhi', 'Agra', 'Mathura'],

    'Category': ['Electronics', 'Clothing', 'Furniture',
                 'Electronics', 'Clothing', 'Furniture',
                 'Electronics', 'Furniture', 'Clothing',
                 'Electronics', 'Clothing', 'Furniture'],

    'Product': ['Laptop', 'Shirt', 'Table', 'Mobile',
                'Jeans', 'Chair', 'Headphones', 'Table',
                'Jacket', 'Mobile', 'Shirt', 'Chair'],

    'Quantity': [2, 3, 1, 2, 4, 1, 2, 3, 2, 1, 5, 2],

    'Unit_Price': [50000, 1200, 15000, 20000,
                   1800, 8000, 2500, 15000,
                   2500, 20000, 1200, 8000],

    'Payment_Mode': ['UPI', 'Card', 'Cash', 'UPI',
                     'Card', 'UPI', 'Cash', 'Card',
                     'UPI', 'Cash', 'Card', 'UPI']
}

df = pd.DataFrame(data)

df['Sales'] = df['Quantity'] * df['Unit_Price']

print(df)
```

Group by single column

```python
city_sales = df.groupby('City')['Sales'].sum()
print(city_sales)
```

Grouping by Multiple Column

```python
city_category_sales = df.groupby(['City', 'Category'])['Sales'].sum()
print(city_category_sales)
```

Flatteing the grouped result

```python
city_category_tables = df.groupby(['City', 'Category'])['Sales'].sum().reset_index()
print(city_category_tables)
```

```python
print(df.groupby(['City', 'Category'])['Sales'].mean())
```

```python
print(df.groupby(['City', 'Category'])['Sales'].agg(['min', 'max']))
```

```python
print(df.groupby('City').size())
```

```python
print(df.groupby('City').count())
```

```python
print(df.groupby('City').nunique())
```

Multiple Aggregation with agg()

```python
sales_report = df.groupby('City')['Sales'].agg(['sum', 'mean', 'min', 'max'])
sales_report
```

```python
category_report = df.groupby('Category').agg(
    Total_Sales=('Sales', 'sum'),
    Average_Price=('Sales', 'mean'),
    Total_quantity=('Quantity', 'sum'),
    Number_of_orders=('Order_ID', 'count')
)
print(category_report)
```

Custom Aggregation

```python
def sales_range(x):
  return x.max() - x.min()

sales_range_report = df.groupby('City')['Sales'].agg(sales_range)
print(sales_range_report)
```

```python
range_report = df.groupby('Category')['Sales'].agg(
    lambda x: x.max() - x.min()
)
range_report
```

Applying Operations to Groups with apply()

```python
top_orders = df.groupby('Customer').apply(
    lambda x: x.nlargest(2, 'Sales')
)
print(top_orders)
```

Group Transformation with transform()

```python
import numpy as np
df['City_Avg_Sales'] = df.groupby('City')['Sales'].transform('mean')
df['Performance'] = np.where(df['Sales'] > df['City_Avg_Sales'], 'Above Average', 'Below Average')
print(df)
```

```python
city_mean = df.groupby('City')['Sales'].transform('mean')
city_std = df.groupby('City')['Sales'].transform('std')
df['Z_Score'] = (df['Sales'] - city_mean) / city_std
print(df)
```

Filtering Group with filter()

```python
large_order = df.groupby('City').filter(
    lambda x: x['Sales'].max() > 50000)
print(large_order)
```

```python
high_sales = df.groupby('City').filter(
    lambda x: x['Sales'].sum() > 50000)
print(high_sales)
```
